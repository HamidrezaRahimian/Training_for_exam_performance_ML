```python
# 01 - Imports and Global Settings

# Core data libraries
import numpy as np  # numerical computations
import pandas as pd  # tabular data handling

# Visualization libraries
import matplotlib.pyplot as plt  # plotting
import seaborn as sns  # statistical plots built on top of matplotlib

# Scikit-learn core utilities
from sklearn.model_selection import (
    train_test_split,  # train test split
    StratifiedKFold,   # stratified cross validation
    GridSearchCV       # hyperparameter tuning with grid search
)

# Scikit-learn pipeline and column transformer
from sklearn.pipeline import Pipeline  # main pipeline object
from sklearn.compose import ColumnTransformer  # different transforms per column group

# Scikit-learn preprocessing and imputation
from sklearn.impute import SimpleImputer, KNNImputer  # imputers for missing values
from sklearn.preprocessing import (
    StandardScaler,    # scaling to zero mean and unit variance
    MinMaxScaler,      # scaling to [0, 1] range (optional)
    OneHotEncoder,     # one hot encoding for categorical features
    OrdinalEncoder     # ordinal encoding (not used here, but imported as option)
)

# Scikit-learn classifiers
from sklearn.neighbors import KNeighborsClassifier  # KNN classifier
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier  # tree based models

# Scikit-learn metrics
from sklearn.metrics import (
    accuracy_score,          # accuracy
    precision_score,         # precision
    recall_score,            # recall
    f1_score,                # F1 score
    roc_auc_score,           # ROC AUC
    roc_curve,               # ROC curve points
    precision_recall_curve,  # precision recall curve points
    confusion_matrix,        # confusion matrix
    classification_report    # detailed classification report
)

# Scikit-learn base classes for custom transformers
from sklearn.base import BaseEstimator, TransformerMixin  # custom transformer base classes

# Statistical tests
from scipy import stats  # chi square or fisher tests

# Warnings
import warnings  # for suppressing non critical warnings

# Global random state for reproducibility
RANDOM_STATE = 42  # global seed used everywhere

# Configure plotting style
sns.set(style="whitegrid", context="notebook")  # seaborn style
plt.rcParams["figure.figsize"] = (8, 5)  # default figure size
plt.rcParams["axes.titlesize"] = 12  # title font size
plt.rcParams["axes.labelsize"] = 11  # axis label font size

# Turn off non essential warnings to keep notebook output clean
warnings.filterwarnings("ignore")  # suppress warnings that are not critical

print("Libraries imported and global settings configured.")
```
